```python
a = pd.read_csv("/content/drive/MyDrive/Data.csv")
a
```

```python
x = a.iloc[:,0:3].values #.iloc[row,col] 0: 3-1
y = a.iloc[:,-1].values
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(x[:,1:3])
x[:,1:3] = imputer.transform(x[:,1:3])
x
```

```python
#text to numeric 1) onehot Ecoding(multiple(>2) Catogary) 2)level Ecoding(less(1,2) Catogary)
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
c = ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[0])],remainder='passthrough')
x = np.array(c.fit_transform(x))
x
```

```python
from sklearn.preprocessing import LabelEncoder
l = LabelEncoder()
y = l.fit_transform(y)
y
```

```python
#spliting data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=1) #Random_state is used for split data randomly
```
